# Week 7: Sales Data Analysis & Dashboard
## End-to-End Analysis Pipeline
This notebook demonstrates Steps 3 through 7: data cleaning, basic statistics, category analysis, monthly sales trends, customer behavior, forecasting, visualizations, and reporting.

```python
import os
import sys
sys.path.append(os.path.abspath('..'))

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

from sales_analyzer import DataLoader, DataCleaner, SalesAnalyzer, SalesVisualizer, ReportGenerator
print('Sales Analyzer package imported successfully.')
```

### 1. Data Cleaning & Preprocessing (Step 3)
Handling missing values (median for numerical, mode for categorical), removing duplicates, and casting datatypes.

```python
loader = DataLoader('../data/raw/sales_data.csv')
df_raw = loader.load_csv()
cleaner = DataCleaner(df_raw)
df_clean = cleaner.clean()
print('Cleaned dataset shape:', df_clean.shape)
cleaner.export_cleaned_data('../data/processed/sales_data_cleaned.csv')
```

### 2. Basic Analysis (Step 4)
Calculating key metrics: total revenue, order count, average order value, unique customers, and date range.

```python
analyzer = SalesAnalyzer(df=df_clean)
stats = analyzer.calculate_basic_stats()
for k, v in stats.items():
    print(f'{k}: {v}')
```

### 3. Category Sales Performance
Ranking categories by total revenue, order count, and revenue share.

```python
category_df = analyzer.analyze_sales_by_category()
category_df
```

### 4. Monthly Sales Trends & Growth (Step 5)
Analyzing sales by month and calculating Month-over-Month (MoM) growth rates.

```python
monthly_df = analyzer.analyze_monthly_trends()
monthly_df
```

### 5. Customer Purchase Patterns & Cohort Analysis
Analyzing repeat buyer behavior, customer lifetime value (CLV), and revenue concentration.

```python
cust_patterns = analyzer.analyze_customer_purchase_patterns()
print('Customer Insights:')
for k, v in cust_patterns.items():
    print(f'  {k}: {v}')

clv = analyzer.calculate_customer_lifetime_value()
print('
Customer Lifetime Value (CLV):')
for k, v in clv.items():
    print(f'  {k}: {v}')

cohort_retention = analyzer.perform_cohort_analysis()
cohort_retention.head()
```

### 6. Sales Forecasting (Simple Moving Average)
Projecting sales for the next 3 months using rolling averages.

```python
forecast = analyzer.generate_sales_forecast(periods=3, window=3)
forecast
```

### 7. Visualizations & Executive Dashboard (Step 6)
Generating monthly trends, category bars, revenue share pie chart, and the 4-panel executive dashboard.

```python
visualizer = SalesVisualizer()
visualizer.generate_all_visualizations(analyzer, output_dir='../data/reports/charts')
fig = visualizer.create_dashboard(analyzer)
plt.show()
```

### 8. Executive Reporting (Step 7)
Generating formatted ASCII summary and exporting full reports to Excel, CSV, and HTML.

```python
reporter = ReportGenerator(analyzer)
print(reporter.generate_text_report())
reporter.export_to_excel(output_path='../data/reports/sales_report.xlsx')
reporter.export_to_csv(output_dir='../data/reports')
reporter.generate_html_report(output_path='../data/reports/sales_report.html')
```
