## Sales Dataset Exploration
This notebook demonstrates Step 2 of the Week 7 curriculum: loading sales data, inspecting dataset structure, assessing data quality, and identifying types.

```python
import os
import sys
sys.path.append(os.path.abspath('..'))

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

from sales_analyzer.data_loader import DataLoader
print('Libraries imported successfully.')
```

### 1. Data Loading
Loading the raw sales dataset from `../data/raw/sales_data.csv`.

```python
data_path = '../data/raw/sales_data.csv'
loader = DataLoader(data_path)
df = loader.load_csv()
df.head()
```

### 2. Dataset Structure and Info
Inspecting data types, non-null counts, and shape.

```python
print(f'Dataset Shape: {df.shape[0]} rows, {df.shape[1]} columns')
df.info()
```

### 3. Summary Statistics
Descriptive statistics for numerical columns (`quantity`, `unit_price`, `total_amount`).

```python
df.describe()
```

### 4. Data Quality Check: Missing Values & Duplicates
Checking for null values across all columns and duplicate order entries.

```python
quality_report = loader.check_data_quality(df)
print('Columns with missing values:', quality_report['columns_with_missing_values'])
print('Duplicate rows:', quality_report['duplicate_rows'])
```

### 5. Categorical Breakdown
Exploring distinct values across categories, regions, and payment methods.

```python
print('Product Categories:
', df['category'].value_counts())
print('
Geographical Regions:
', df['region'].value_counts())
print('
Payment Methods:
', df['payment_method'].value_counts())
```

### 6. Initial Exploratory Plots
Histogram of total order amounts and orders per category.

```python
fig, axes = plt.subplots(1, 2, figsize=(14, 5))
df['category'].value_counts().plot(kind='bar', ax=axes[0], color='#2563EB', title='Order Count by Category')
axes[0].set_ylabel('Number of Orders')
axes[0].tick_params(axis='x', rotation=45)

df['total_amount'].plot(kind='hist', bins=30, ax=axes[1], color='#10B981', title='Order Value Distribution', edgecolor='black')
axes[1].set_xlabel('Order Value in USD')
plt.tight_layout()
plt.show()
```
